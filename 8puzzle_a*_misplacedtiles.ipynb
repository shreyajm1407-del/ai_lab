```python
import heapq

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)   # 0 = blank


def misplaced(state):
    return sum(state[i] != goal[i] and state[i] != 0
               for i in range(9))


def children(state):
    pos = state.index(0)
    row, col = divmod(pos, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_pos = r * 3 + c
            new_state = list(state)

            new_state[pos], new_state[new_pos] = \
                new_state[new_pos], new_state[pos]

            yield tuple(new_state)


def a_star(start):
    # (f, h, state, path)
    open_list = []
    h = misplaced(start)

    heapq.heappush(open_list, (h, h, start, [start]))
    closed = set()

    while open_list:
        f, h, state, path = heapq.heappop(open_list)

        if state in closed:
            continue

        closed.add(state)

        if state == goal:
            return path

        for child in children(state):
            if child not in closed:
                g = len(path)
                h = misplaced(child)
                f = g + h

                heapq.heappush(
                    open_list, (f, h, child, path + [child])
                )


start = (2, 8, 3,
         1, 6, 4,
         0, 7, 5)

solution = a_star(start)

print("Solution:")

for state in solution:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```
