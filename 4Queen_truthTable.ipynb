```python

def cost(s):
    return sum(
        s[i] == s[j] or abs(s[i]-s[j]) == j-i
        for i in range(4) for j in range(i+1, 4)
    )

def hill_climb(s):
    while cost(s) > 0:
        neighbors = [
            s[:i] + [r] + s[i+1:]
            for i in range(4) for r in range(4)
            if r != s[i]
        ]
        best = min(neighbors, key=cost)
        if cost(best) >= cost(s):
            return "Local minimum reached", s
        s = best
    return "Goal reached", s

print(hill_climb([1, 3, 0, 2]))
```
