```python
# TIC-TAC-TOE USING GAME TREE AND MINIMAX
# Human vs Computer

# -----------------------------
# Display the board
# -----------------------------
def display_board(board):
    print()
    print("-------------")
    for i in range(3):
        print("|", board[i][0], "|", board[i][1], "|", board[i][2], "|")
        print("-------------")
    print()


# -----------------------------
# Check whether a player wins
# -----------------------------
def check_winner(board, player):
    # Rows
    for row in board:
        if row[0] == player and row[1] == player and row[2] == player:
            return True

    # Columns
    for col in range(3):
        if board[0][col] == player and \
           board[1][col] == player and \
           board[2][col] == player:
            return True

    # Diagonal 1
    if board[0][0] == player and \
       board[1][1] == player and \
       board[2][2] == player:
        return True

    # Diagonal 2
    if board[0][2] == player and \
       board[1][1] == player and \
       board[2][0] == player:
        return True

    return False


# -----------------------------
# Check whether board is full
# -----------------------------
def board_full(board):
    for row in board:
        for cell in row:
            if cell == " ":
                return False
    return True


# -----------------------------
# Minimax Algorithm
# -----------------------------
def minimax(board, computer, human, is_maximizing):

    # Computer wins
    if check_winner(board, computer):
        return 1

    # Human wins
    if check_winner(board, human):
        return -1

    # Draw
    if board_full(board):
        return 0

    # Computer's turn - MAXIMIZE
    if is_maximizing:
        best_score = -float("inf")

        for i in range(3):
            for j in range(3):
                if board[i][j] == " ":

                    # Make computer move
                    board[i][j] = computer

                    # Explore future moves
                    score = minimax(
                        board,
                        computer,
                        human,
                        False
                    )

                    # Undo move
                    board[i][j] = " "

                    best_score = max(best_score, score)

        return best_score

    # Human's turn - MINIMIZE
    else:
        best_score = float("inf")

        for i in range(3):
            for j in range(3):
                if board[i][j] == " ":

                    # Make human move
                    board[i][j] = human

                    # Explore future moves
                    score = minimax(
                        board,
                        computer,
                        human,
                        True
                    )

                    # Undo move
                    board[i][j] = " "

                    best_score = min(best_score, score)

        return best_score


# -----------------------------
# Find Computer's Best Move
# -----------------------------
def computer_move(board, computer, human):

    best_score = -float("inf")
    best_move = None

    # Generate all possible computer moves
    for i in range(3):
        for j in range(3):

            if board[i][j] == " ":

                # Try this move
                board[i][j] = computer

                # Apply Minimax
                score = minimax(
                    board,
                    computer,
                    human,
                    False
                )

                # Undo move
                board[i][j] = " "

                # Choose maximum score
                if score > best_score:
                    best_score = score
                    best_move = (i, j)

    return best_move


# -----------------------------
# Main Game
# -----------------------------
def play_game():

    # Initialize empty board
    board = [
        [" ", " ", " "],
        [" ", " ", " "],
        [" ", " ", " "]
    ]

    print("================================")
    print("       TIC-TAC-TOE GAME")
    print("================================")

    # Human chooses X or O
    while True:
        human = input("Choose your symbol (X/O): ").upper()

        if human == "X" or human == "O":
            break

        print("Invalid choice. Please enter X or O.")

    # Computer gets the other symbol
    if human == "X":
        computer = "O"
    else:
        computer = "X"

    print("Human:", human)
    print("Computer:", computer)

    print("\nCell numbers:")
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")

    # Human always starts
    while True:

        # -------------------------
        # Human Move
        # -------------------------
        display_board(board)

        while True:
            try:
                position = int(input("Enter an empty cell (1-9): "))

                if position < 1 or position > 9:
                    print("Please enter a number from 1 to 9.")
                    continue

                row = (position - 1) // 3
                col = (position - 1) % 3

                if board[row][col] != " ":
                    print("That cell is already occupied.")
                    continue

                # Place human symbol
                board[row][col] = human
                break

            except ValueError:
                print("Please enter a valid number.")

        # Check Human win
        if check_winner(board, human):
            display_board(board)
            print("Human Wins!")
            break

        # Check Draw
        if board_full(board):
            display_board(board)
            print("Game Draw!")
            break

        # -------------------------
        # Computer Move
        # -------------------------
        print("Computer is thinking...")

        move = computer_move(board, computer, human)

        if move is not None:
            row, col = move
            board[row][col] = computer

        print("Computer placed", computer, "at cell",
              row * 3 + col + 1)

        # Check Computer win
        if check_winner(board, computer):
            display_board(board)
            print("Computer Wins!")
            break

        # Check Draw
        if board_full(board):
            display_board(board)
            print("Game Draw!")
            break


# -----------------------------
# Start the game
# -----------------------------
play_game()
```
